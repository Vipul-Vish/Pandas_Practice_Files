#### GroupBy & Aggregation

```python
data = {
    "Name": ["Vipul", "Ram", "Musu", "Lalu", "Riya", "Aman"],
    "City": ["Delhi", "Mumbai", "Delhi", "Mumbai", "Delhi", "Mumbai"],
    "Marks": [85, 70, 92, 65, 95, 78],
    "Age": [20, 25, 22, 25, 19, 21]
}

df = pd.DataFrame(data)
```

```python
df.groupby("City")["Marks"].mean()
```

```python
df.groupby("City")["Name"].count()
```

```python
df.groupby("City")["Marks"].max()
```

```python
df.groupby("City")["Marks"].agg(
    ["mean", "max", "min", "count"]
)
```

```python
df["City"].value_counts()
```

```python
df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Vipul", "Ram", "Musu"]
})

df2 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [85, 70, 92]
})

print(df1)
print(df2)
```

```python
result = pd.merge(df1, df2, on="ID")
print(result)
```

```python
pd.merge(df1, df2, on="ID", how="inner")
pd.merge(df1, df2, on="ID", how="left")
pd.merge(df1, df2, on="ID", how="right")
pd.merge(df1, df2, on="ID", how="outer")
```

```python
batch1 = pd.DataFrame({
    "Name": ["Vipul", "Ram"],
    "Marks": [85, 70]
})

batch2 = pd.DataFrame({
    "Name": ["Musu", "Lalu"],
    "Marks": [92, 65]
})

result = pd.concat([batch1, batch2], ignore_index=True)

print(result)
```

```python
df1 = pd.DataFrame({
    "Name": ["Vipul", "Ram", "Musu"]
}, index=[1, 2, 3])

df2 = pd.DataFrame({
    "Marks": [85, 70, 92]
}, index=[1, 2, 3])

result = df1.join(df2)
print(result)
```
