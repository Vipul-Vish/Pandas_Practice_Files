## Business Insights & Final Report

### Task 1: Overall Business Performance
#### Question: Calculate Total Sales, Total Profit, Total Orders, Total Unique Customers, and Overall Profit Margin.

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()

profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Orders:", total_orders)
print("Total Unique Customers:", total_customers)
print("Overall Profit Margin:", round(profit_margin, 2), "%")
```

### Task 2: Best and Worst Performing Category
#### Question: Identify the category with the highest Sales, highest Profit, and lowest Profit.

```python
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

print(category_analysis)

best_sales_category = category_analysis["Total_Sales"].idxmax()
best_profit_category = category_analysis["Total_Profit"].idxmax()
worst_profit_category = category_analysis["Total_Profit"].idxmin()

print("Highest Sales Category:", best_sales_category)
print("Highest Profit Category:", best_profit_category)
print("Lowest Profit Category:", worst_profit_category)
```

### Task 3: Regional Performance
#### Question: Calculate total Sales, total Profit, and Profit Margin for each Region. Identify the regions with the highest Sales and Profit.

```python
region_analysis = df.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

region_analysis["Profit_Margin"] = (
    region_analysis["Total_Profit"] /
    region_analysis["Total_Sales"]
) * 100

print(region_analysis.round(2))

highest_sales_region = region_analysis["Total_Sales"].idxmax()
highest_profit_region = region_analysis["Total_Profit"].idxmax()

print("Highest Sales Region:", highest_sales_region)
print("Highest Profit Region:", highest_profit_region)
```

### Task 4: Top 10 Loss-Making Products
#### Question: Find the 10 products with the lowest total Profit. Display Product Name, Total Sales, and Total Profit.

```python
product_analysis = df.groupby("Product Name").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

loss_making_products = product_analysis.sort_values(
    by="Total_Profit",
    ascending=True
).head(10)

print(loss_making_products)
```

### Task 5: Monthly Sales Trend
#### Question: Calculate total Sales for each month across all years. Identify the months with the highest and lowest Sales.

```python
import numpy as np
df["Order Date"] = pd.to_datetime(df["Order Date"])

df["Order_Month"] = df["Order Date"].dt.month

monthly_sales = df.groupby("Order_Month")["Sales"].sum()
monthly_sales = monthly_sales.sort_index()

print("Monthly Sales:")
print(monthly_sales)

highest_sales_month = monthly_sales.idxmax()
lowest_sales_month = monthly_sales.idxmin()

print("Month with Highest Sales:", highest_sales_month)
print("Month with Lowest Sales:", lowest_sales_month)
```

### Task 6: Discount vs Profit Analysis
#### Question: Divide orders into No Discount, Low Discount, and High Discount groups. Calculate Total Sales, Total Profit, and Average Profit for each group.

```python
conditions = [
    df["Discount"] == 0,
    (df["Discount"] > 0) & (df["Discount"] <= 0.2),
    df["Discount"] > 0.2
]

labels = [
    "No Discount",
    "Low Discount",
    "High Discount"
]

df["Discount_Group"] = np.select(
    conditions,
    labels,
    default="Other"
)

discount_analysis = df.groupby("Discount_Group").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Average_Profit=("Profit", "mean")
)

print(discount_analysis.round(2))
```
